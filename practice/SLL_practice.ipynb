### Linked List

```python
class Node:
    def __init__(self, value: int):
        self.value: int = value
        self.next: Node | None = None
```

```python
def printNodes(head: Node):
    count = 0
    curr = head

    print("[OP] ", end='')
    while curr:
        print(f'{curr.value}', end='')
        count += 1
        if curr.next:
            print(' > ', end='')
        curr = curr.next
    print(f'\nNo. of nodes: {count}\n')
```

```python
def inputHelper(type: int):
    '''
    1 for value and 0 for position
    '''
    inp = None
    while not inp:
        try:
            if type:
                inp = int(input("[I/O] Enter the value for the new node: "))
            else:
                inp = int(input("[I/O] Enter the position of the new node: "))
        except ValueError:
            print("[ERROR] No Input Detected.")
            continue
    return inp

def insertAtFirst(head: Node, val: int) -> Node:
    print(f"[OP] Inserting new node with {val} before head.")

    newNode = Node(val)
    newNode.next = head
    head = newNode
    print(f"[SUCCESS] Operation finished successfully.")
    printNodes(head)
    return head

def insertAtPos(head: Node, pos: int, val: int) -> Node: #O(N)
    print(f"[OP] Inserting new node with {val} at {pos}th position.")

    if pos < 1:
        print("[ERROR] Position must be greater than 1")
        return head
    if pos == 1:
        return insertAtFirst(head, val)

    newNode = Node(val)
    curr = head

    for i in range(1, pos - 1):
        if not curr.next:
            print(f"[INTERRUPT] {i}th node is the last node.")
            return insertAtEnd(head, val)
        curr = curr.next

    newNode.next = curr.next
    curr.next = newNode
    print(f"[SUCCESS] Operation finished successfully.")
    printNodes(head)
    return head

def insertAtEnd(head: Node, val: int):
    print(f"[OP] Inserting new node with {val} at the end.")

    curr = head
    while curr.next:
        curr = curr.next

    newNode = Node(val)
    curr.next = newNode
    print(f"[SUCCESS] Operation finished successfully.")
    printNodes(head)
    return head
```

```python
# Manual Trial [WORKS]
head = Node(50)
head.next = Node(70)
head.next.next = Node(90)

printNodes(head)

head = insertAtPos(head, 2, 30)
printNodes(head)
```
